## 01 - Análise Exploratória de Dados (EDA)

Sistema de apoio ao diagnóstico de câncer de mama a partir do **Wisconsin Diagnostic Breast Cancer (WDBC)**.

O dataset reúne medidas extraídas de imagens digitalizadas de massas mamárias. Cada amostra descreve características dos núcleos celulares (raio, textura, perímetro, área, suavidade etc.) calculadas como **média**, **erro padrão** e **pior valor**, totalizando 30 features após a limpeza.

**Objetivo desta etapa:** entender a estrutura dos dados, tratar colunas irrelevantes, avaliar o balanceamento das classes, analisar correlações e exportar uma base limpa para a modelagem.

## 1. Carga e visão geral dos dados

## 2. Limpeza: remoção de colunas irrelevantes

Duas colunas não carregam informação útil para o modelo:

- **`id`**: identificador único do paciente, sem valor preditivo.
- **`Unnamed: 32`**: coluna-artefato gerada na exportação do CSV (totalmente vazia / `NaN`).

```python
# Remove o identificador (sem valor preditivo) e a coluna-artefato vazia ("Unnamed: 32")
df_clean = df.drop(columns=["id", "Unnamed: 32"])
```

```python
df_clean.info()
```

## 3. Codificação da variável-alvo

Mapeamos o diagnóstico para valores numéricos, definindo **maligno como classe positiva (1)** - a classe de maior interesse clínico, já que falsos negativos são o erro mais grave neste contexto.

- `M` (maligno) → `1`
- `B` (benigno) → `0`

```python
# Maligno = 1 (classe positiva, de maior interesse clínico) | Benigno = 0
df_clean["diagnosis"] = df_clean["diagnosis"].map({"M": 1, "B": 0})
```

Confirmando a proporção das classes: aproximadamente **63% benignas** e **37% malignas** - desbalanceamento moderado, que será tratado com *split* estratificado na etapa de modelagem.

```python
df_clean["diagnosis"].value_counts()
```

## 4. Análise de correlação

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
correlation_matrix = df_clean.corr()
```

Visão geral da matriz de correlação completa:

```python
sns.heatmap(correlation_matrix)
plt.show()
```

Correlação de cada feature com o diagnóstico:

```python
correlation_matrix["diagnosis"].sort_values()
```

As features de **média** mais relacionadas ao diagnóstico (`radius_mean`, `perimeter_mean`, `area_mean`) descrevem essencialmente o mesmo conceito - o **tamanho** do núcleo - e por isso são altamente redundantes entre si. O heatmap abaixo, restrito às colunas `_mean`, evidencia essa multicolinearidade.

```python
# Isola apenas as features de média para inspecionar a multicolinearidade entre elas
columns_mean = [column for column in df_clean.columns if column.endswith("_mean")]
```

```python
correlation_matrix_mean = df_clean[columns_mean].corr()
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix_mean, annot=True)
plt.title("Correlação entre as features de média (_mean)")
plt.show()
```

## 5. Distribuição das features "campeãs" por diagnóstico

Para reduzir a redundância, selecionamos uma feature representativa por grupo de informação e comparamos sua distribuição entre tumores benignos e malignos. Boa separação entre as classes indica alto poder discriminativo.

```python
# Uma feature representativa por grupo de informação (evita redundância)
columns_to_plot = ["texture_worst", "smoothness_worst", "symmetry_worst", "concave points_worst", "perimeter_worst"]

for column in columns_to_plot:
    plt.figure(figsize=(8, 5))
    sns.histplot(data=df_clean, x=column, hue="diagnosis", kde=True)
    plt.title(f"Distribuição de {column} por tipo de diagnóstico")
    plt.show()
```

## 6. Exportação do dataset processado

A base limpa e codificada é salva em `wdbc_processed.csv`, que serve de entrada para o notebook de modelagem (`02-modelagem.ipynb`).

```python
# Exporta a base limpa e codificada como entrada para o notebook de modelagem
df_clean.to_csv("../data/processed/wdbc_processed.csv", encoding="utf-8", index=False)
```

## Conclusões da EDA

- O dataset tem **569 amostras** e **30 features** após a remoção de `id` e `Unnamed: 32`.
- As classes estão **moderadamente desbalanceadas** (~63% benignas / ~37% malignas) → usar *split* estratificado.
